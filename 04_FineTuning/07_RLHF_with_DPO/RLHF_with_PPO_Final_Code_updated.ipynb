## 2. Import Libraries
Import the necessary modules for data handling, reward modeling, and PPO training.


```python
# =========================================================
# 1) IMPORTS
# =========================================================
import os
import gc
import random
import torch
import trl

from datasets import Dataset
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    AutoModelForSequenceClassification,
)

from trl import RewardTrainer, RewardConfig
from trl.experimental.ppo import PPOTrainer, PPOConfig
```

```python
print("TRL version:", trl.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

## 3. Global Settings
Configure memory allocations and set random seeds for reproducibility.


```python
# =========================================================
# 2) GLOBAL SETTINGS
# =========================================================

os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

SEED = 42
random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

dtype = torch.float16 if torch.cuda.is_available() else torch.float32
```

```python
def clear_memory():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
```

## 4. Model Selection
Define the base model to be used for both the Reward Model and the Policy Model.


```python
# =========================================================
# 3) MODEL NAME
# =========================================================

# model_name = "Qwen/Qwen2.5-0.5B-Instruct"

model_name = "HuggingFaceTB/SmolLM2-135M-Instruct"
```

## 5. Reward Model Tokenizer
Load the tokenizer for the Reward Model. Right padding is typically preferred for sequence classification (Reward Modeling).


```python
# =========================================================
# 4) REWARD MODEL TOKENIZER
# =========================================================

tok_rm = AutoTokenizer.from_pretrained(model_name)

if tok_rm.pad_token is None:
    tok_rm.pad_token = tok_rm.eos_token

tok_rm.padding_side = "right"
```

## 6. Reward Dataset
Create a preference dataset consisting of a prompt, a chosen response, and a rejected response.


```python
# =========================================================
# 5) REWARD DATASET
# =========================================================
# We create explicit preference format:
# prompt + chosen
# prompt + rejected

raw_reward_data = [
    {
        "prompt": "How should I prepare for exams?",
        "chosen": "Make a schedule, revise daily, practice previous questions, and sleep well.",
        "rejected": "Do nothing and panic at the end.",
    },
    {
        "prompt": "How do I improve fitness?",
        "chosen": "Exercise consistently, eat balanced meals, hydrate well, and recover properly.",
        "rejected": "Starve yourself, skip sleep, and exercise randomly.",
    },
    {
        "prompt": "What is AI?",
        "chosen": "AI is the ability of machines to perform tasks that usually require human intelligence.",
        "rejected": "AI is random magic with no real meaning.",
    },
    {
        "prompt": "How can I improve my teaching skills?",
        "chosen": "Explain concepts step by step, use examples, ask questions, and summarize key points clearly.",
        "rejected": "Speak fast, ignore students, and make the topic unnecessarily complicated.",
    },
]
```

```python
def make_chat_prompt(user_prompt):
    messages = [
        {"role": "user", "content": user_prompt}
    ]

    return tok_rm.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
    )
```

```python
def make_reward_example(example):
    return {
        "prompt": make_chat_prompt(example["prompt"]),
        "chosen": example["chosen"] + tok_rm.eos_token,
        "rejected": example["rejected"] + tok_rm.eos_token,
    }
```

```python
reward_data = [make_reward_example(x) for x in raw_reward_data]
```

```python
reward_ds = Dataset.from_list(reward_data)
```

## 7. Load Reward Model
Load a sequence classification model to act as our Reward Model. It will output a single scalar reward for a given sequence.


```python
# =========================================================
# 6) LOAD REWARD MODEL
# =========================================================

# Load the model for sequence classification (num_labels=1 outputs a single reward score)
rm = AutoModelForSequenceClassification.from_pretrained(
    model_name,
    num_labels=1,
    torch_dtype=dtype,
)

rm.config.pad_token_id = tok_rm.pad_token_id
rm.config.use_cache = False

if hasattr(rm, "gradient_checkpointing_enable"):
    rm.gradient_checkpointing_enable()
```

## 8. Reward Training Config
Configure the hyperparameters for training the Reward Model.


```python
# 7) REWARD TRAINING CONFIG
# =========================================================

rm_args = RewardConfig(
    output_dir="./rm_out",

    per_device_train_batch_size=1,
    per_device_eval_batch_size=1,

    num_train_epochs=1,
    learning_rate=1e-5,

    logging_steps=1,
    eval_strategy="no",
    save_strategy="no",
    report_to="none",

    fp16=False,
    bf16=False,

    max_length=256,
    gradient_checkpointing=True,
    remove_unused_columns=False,
)
```

## 9. Train Reward Model
Initialize the `RewardTrainer` and train the model to distinguish between chosen and rejected responses.


```python
# =========================================================
# 8) TRAIN REWARD MODEL
# =========================================================

# Initialize RewardTrainer
rm_trainer = RewardTrainer(
    model=rm,
    args=rm_args,
    processing_class=tok_rm,
    train_dataset=reward_ds,
)
```

```python
print("=== Training Reward Model ===")
rm_trainer.train()

rm_trainer.save_model("./rm_out")
tok_rm.save_pretrained("./rm_out")

print("Reward Model saved at ./rm_out")
```

## 10. Clear Memory Before PPO
Free up GPU memory by deleting the reward trainer and model. We will load the trained reward model back later in inference mode.


```python
# =========================================================
# 9) CLEAR MEMORY BEFORE PPO
# =========================================================

del rm
del rm_trainer
del tok_rm
clear_memory()
```

## 11. PPO Tokenizer
Load the tokenizer for PPO generation. Left padding is required for auto-regressive text generation tasks.


```python
# =========================================================
# 10) PPO TOKENIZER
# =========================================================

tok = AutoTokenizer.from_pretrained(model_name)

if tok.pad_token is None:
    tok.pad_token = tok.eos_token

# Left padding is strictly required for causal language modeling generation
tok.padding_side = "left"
```

## 12. Load PPO Models (Policy & Reference)
PPO requires a trainable Policy model and a frozen Reference model (to compute the KL penalty and prevent the policy from diverging too far).


```python
# =========================================================
# 11) LOAD PPO MODELS
# =========================================================

# Policy model - this model will be updated by PPO
policy = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=dtype,
)

policy.config.pad_token_id = tok.pad_token_id
policy.config.use_cache = False

if hasattr(policy, "gradient_checkpointing_enable"):
    policy.gradient_checkpointing_enable()

if hasattr(policy, "enable_input_require_grads"):
    policy.enable_input_require_grads()
```

The Reference Model is a frozen copy of the original Policy Model. It is never trained during PPO. Its only purpose is to measure how much the current Policy Model has changed and apply a KL penalty if it drifts too far from its original behavior.

```python
# Reference model - fixed copy for KL penalty
ref_model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=dtype,
)

ref_model.config.pad_token_id = tok.pad_token_id
ref_model.config.use_cache = False
ref_model.eval()

##trainig is not require
for param in ref_model.parameters():
    param.requires_grad_(False)
```

## 13. Load Reward Model for PPO
Load the reward model we just trained. It will be kept frozen and used to score the responses generated by the Policy model.


```python
# Reward model - trained above, fixed during PPO
reward_model = AutoModelForSequenceClassification.from_pretrained(
    "./rm_out",
    num_labels=1,
    torch_dtype=dtype,
)

reward_model.config.pad_token_id = tok.pad_token_id
reward_model.config.use_cache = False
reward_model.eval()

##trainig is not require
for param in reward_model.parameters():
    param.requires_grad_(False)
```

## 14. Load Value Model for PPO
The Value model is used in PPO to estimate the expected return of a state. It is trainable and updated alongside the policy.


```python
# Value model - trainable value head for PPO
value_model = AutoModelForSequenceClassification.from_pretrained(
    "./rm_out",
    num_labels=1,
    torch_dtype=dtype,
)

value_model.config.pad_token_id = tok.pad_token_id
value_model.config.use_cache = False

if hasattr(value_model, "gradient_checkpointing_enable"):
    value_model.gradient_checkpointing_enable()
```

```python
clear_memory()
```

## 15. PPO Prompt Dataset
Create a dataset of prompts that the Policy model will try to answer. The Reward model will score its answers.


```python
# =========================================================
# 12) PPO PROMPT DATASET
# =========================================================

prompt_ds = Dataset.from_list([
    {"prompt": "How should I prepare for exams?"},
    {"prompt": "How do I improve fitness?"},
    {"prompt": "What is AI?"},
    {"prompt": "How can I improve my teaching skills?"},
])
```

```python
def make_ppo_prompt(user_prompt):
    messages = [
        {"role": "user", "content": user_prompt}
    ]

    return tok.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
    )
```

```python
def tokenize_for_ppo(example):
    prompt_text = make_ppo_prompt(example["prompt"])

    encoded = tok(
        prompt_text,
        padding=False,
        truncation=True,
        max_length=128,
    )

    return {
        "input_ids": encoded["input_ids"],
        "length": len(encoded["input_ids"]),
    }
```

```python
ppo_train_dataset = prompt_ds.map(
    tokenize_for_ppo,
    remove_columns=prompt_ds.column_names,
)
```

```python
ppo_train_dataset = ppo_train_dataset.filter(lambda x: x["length"] <= 128)
```

```python
print("PPO dataset sample:")
print(ppo_train_dataset[0])
```

## 16. PPO Config
Set up the hyperparameters for the PPO training algorithm.


```python
#=========================================================
# 13) PPO CONFIG
# =========================================================

ppo_args = PPOConfig(
    output_dir="./ppo_out",

    per_device_train_batch_size=1,
    gradient_accumulation_steps=1,

    learning_rate=1e-6,

    num_mini_batches=1,
    num_ppo_epochs=1,

    total_episodes=6,
    response_length=16,

    logging_steps=1,
    save_strategy="no",
    report_to="none",

    fp16=False,
    bf16=False,

    gradient_checkpointing=True,
    remove_unused_columns=False,

    seed=SEED,
)
```

## 17. PPO Trainer
Initialize the `PPOTrainer` containing all 4 models: Policy, Reference, Reward, and Value.


```python
# =========================================================
# 14) PPO TRAINER
# =========================================================

ppo_trainer = PPOTrainer(
    args=ppo_args,
    processing_class=tok,
    model=policy,
    ref_model=ref_model,
    reward_model=reward_model,
    value_model=value_model,
    train_dataset=ppo_train_dataset,
)

print("PPOTrainer initialized successfully!")
```

## 18. Train PPO
Start the PPO training loop. The policy model will generate responses, get scored by the reward model, and update its weights to maximize the reward.


```python
# =========================================================
# 15) TRAIN PPO
# =========================================================

print("=== Training Policy with PPO ===")
ppo_trainer.train()
```

## 19. Save Final Policy
Save the weights of the fine-tuned (aligned) policy model.


```python
# =========================================================
# 16) SAVE FINAL PPO POLICY MODEL
# =========================================================

ppo_trainer.save_model("./ppo_out")
tok.save_pretrained("./ppo_out")

print("Training completed and final policy model saved to ./ppo_out")
```

```python
clear_memory()
```

## 20. Inference
Load the final model and test it with a prompt to see its aligned generation.


```python
# =========================================================
# 17) INFERENCE FROM FINAL PPO MODEL
# =========================================================

from transformers import AutoTokenizer, AutoModelForCausalLM

final_model_path = "./ppo_out"

infer_tok = AutoTokenizer.from_pretrained(final_model_path)

if infer_tok.pad_token is None:
    infer_tok.pad_token = infer_tok.eos_token

# Left padding is strictly required for causal language modeling generation
infer_tok.padding_side = "left"

infer_model = AutoModelForCausalLM.from_pretrained(
    final_model_path,
    torch_dtype=dtype,
    device_map="auto" if torch.cuda.is_available() else None,
)

infer_model.eval()

test_prompt = "How should I prepare for exams?"

messages = [
    {"role": "user", "content": test_prompt}
]

formatted_prompt = infer_tok.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True,
)

inputs = infer_tok(
    formatted_prompt,
    return_tensors="pt",
    padding=True,
    truncation=True,
).to(infer_model.device)

with torch.no_grad():
    generated_ids = infer_model.generate(
        **inputs,
        max_new_tokens=80,
        do_sample=True,
        temperature=0.7,
        top_p=0.9,
        pad_token_id=infer_tok.pad_token_id,
        eos_token_id=infer_tok.eos_token_id,
    )

output = infer_tok.decode(generated_ids[0], skip_special_tokens=True)

print("=== Final Model Output ===")
print(output)
```
